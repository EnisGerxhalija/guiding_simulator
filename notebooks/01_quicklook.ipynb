```python
import matplotlib.pyplot as plt
import numpy as np

from guiding_simulator.config import load_run_config, load_settings
from guiding_simulator.example import sampled_maximum, simulate_peak
```

`load_settings()` finds the project root (and its `.env`) from this folder, so paths work the same as in scripts.

```python
settings = load_settings()
config = load_run_config(settings.config_dir / "default.toml")
config
```

```python
rng = np.random.default_rng(config.seed)
p = config.example
x = np.linspace(p.x_start, p.x_stop, p.n_points)
y = simulate_peak(
    x,
    amplitude=p.amplitude,
    centre=p.centre,
    sigma=p.sigma,
    noise_std=p.noise_std,
    rng=rng,
)
peak = sampled_maximum(x, y)
peak
```

```python
fig, ax = plt.subplots()
ax.plot(x, y, ".", ms=3, label="simulated data")
ax.axvline(
    peak.position, ls="--", c="k", label=f"largest sample, x = {peak.position:.2f}"
)
ax.set(xlabel="x (units)", ylabel="signal (units)")
ax.legend()
# fig.savefig(settings.output_dir / "quicklook.png", dpi=200)
```
